**Document 1**

```
Q: What is the return policy?
A: Items can be returned within 30 days of purchase with original receipt.

Q: Do you offer international shipping?
A: Yes, we ship to over 50 countries worldwide. Shipping times vary by location.

Q: How do I track my order?
A: Use the tracking number sent to your email after shipment.
```

**For semantic meaning in a Q and A, we use paragraph chunking so the Q and A are in the same vector space**

````python
def para_chunk(text):
  # split by each paragraphs (double newlines)
  faqs = text.split('\n\n')

  # print("FAQs: ", faqs)

  chunks = []
  # current_chunk = ""

  for faq in faqs:
    faq = faq.strip()
    if not faq:
      continue
    
    chunks.append(faq)

  return chunks
    
````

````python
faqs = """
  Q: What is the return policy?
  A: Items can be returned within 30 days of purchase with original receipt.

  Q: Do you offer international shipping?
  A: Yes, we ship to over 50 countries worldwide. Shipping times vary by location.

  Q: How do I track my order?
  A: Use the tracking number sent to your email after shipment.
"""
````

````python
chunks = para_chunk(faqs)

for i, chunk in enumerate(chunks):
  print(chunk)
  print("=" * 80)
````
